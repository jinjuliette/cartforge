# 01 - Project setup

## Objective

Set up the Databricks environment and create the schemas used to organize the data pipeline.

The project follows a layered data architecture:

- **Bronze**: raw ingested data
- **Silver**: cleaned and transformed data
- **Gold**: curated data for analytics

All project datasets will be organized under the `fmcg` catalog.

```python
%sql

CREATE CATALOG IF NOT EXISTS fmcg;
USE CATALOG fmcg;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS fmcg.gold;
CREATE SCHEMA IF NOT EXISTS fmcg.silver;
CREATE SCHEMA IF NOT EXISTS fmcg.bronze;
```
